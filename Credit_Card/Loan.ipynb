```python
num_cols = df.select_dtypes(
    include=['int64', 'float64']
).columns

for col in num_cols:
    
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    
    IQR = Q3 - Q1
    
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    
    outliers = df[
        (df[col] < lower) |
        (df[col] > upper)
    ]
    
    print(f"{col}: {len(outliers)} outliers")
```

## Label Encoder 

```python
le=LabelEncoder()
```

```python
df['Employment_Status_Encoder']=le.fit_transform(df['Employment_Status'].astype(str))
df['Marital_Status_Encoder']=le.fit_transform(df['Marital_Status'].astype(str))
df['Loan_Purpose_Encoded'] = le.fit_transform(df['Loan_Purpose'].astype(str))
df['Property_Area_Encoded'] = le.fit_transform(df['Property_Area'].astype(str))
df['Education_Level_Encoded'] = le.fit_transform(df['Education_Level'].astype(str))
df['Gender_Encoded'] = le.fit_transform(df['Gender'].astype(str))
df['Employer_Category_Encoded'] = le.fit_transform(df['Employer_Category'].astype(str))
df['Loan_Approved_Encoded'] = le.fit_transform(df['Loan_Approved'].astype(str))
```

## Train_Test_Split

```python
x=df[['Applicant_Income', 'Coapplicant_Income', 'Employment_Status_Encoder', 'Age', 
        'Marital_Status_Encoder', 'Dependents', 'Credit_Score', 'Existing_Loans', 
        'DTI_Ratio', 'Savings', 'Collateral_Value', 'Loan_Amount', 
        'Loan_Term', 'Loan_Purpose_Encoded', 'Property_Area_Encoded', 'Education_Level_Encoded', 
        'Gender_Encoded', 'Employer_Category_Encoded']]
y=df['Loan_Approved_Encoded']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

## StandStandardScaler

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
x_train_scaled ,x_train_scaled 
```

## Logistic Regression

```python
lr=LogisticRegression(penalty='l2', C=1.0, solver='lbfgs', max_iter=2000, random_state=42)
lr.fit(x_train,y_train)
lr_predict=lr.predict(x_test)
lr_predict
```

## Decision Tree

```python
dt=DecisionTreeClassifier(criterion='gini', max_depth=5, min_samples_split=5, random_state=42)
dt.fit(x_train,y_train)
dt_predict=dt.predict(x_test)
dt_predict
```

## Random Forest

```python
rf = RandomForestClassifier(n_estimators=150, max_depth=8, min_samples_split=4, random_state=42)
rf.fit(x_train, y_train)
rf_predict = rf.predict(x_test)
rf_predict
```

## Support Vector Machine (SVM)

```python
from sklearn.linear_model import SGDClassifier
svm = SGDClassifier(loss='hinge', random_state=42)
svm.fit(x_train_scaled, y_train)
svm_predict = svm.predict(x_test_scaled)
```

## K-Nearest Neighbors (KNN)

```python
knn = KNeighborsClassifier(n_neighbors=5, weights='distance', metric='minkowski', p=2)
knn.fit(x_train_scaled, y_train)
knn_predict = knn.predict(x_test_scaled)
knn_predict 
```

### Naive Baye

```python
nb = GaussianNB(var_smoothing=1e-09)
nb.fit(x_train, y_train)
nb_predict = nb.predict(x_test)
nb_predict
```

### COMPARATIVE BENCHMARKING TABLE

```python
results = {
    'Model': ['Logistic Regression', 'Decision Tree', 'Random Forest', 'Svm', 'KNN', 'Naive Bayes'],
    
    'Accuracy': [
        accuracy_score(y_test, lr_predict),
        accuracy_score(y_test, dt_predict),
        accuracy_score(y_test, rf_predict),
        accuracy_score(y_test, svm_predict),
        accuracy_score(y_test, knn_predict),
        accuracy_score(y_test, nb_predict)
    ],
    
    'Precision': [
        precision_score(y_test, lr_predict, zero_division=0),
        precision_score(y_test, dt_predict, zero_division=0),
        precision_score(y_test, rf_predict, zero_division=0),
        precision_score(y_test, svm_predict, zero_division=0),  # 👈 આ લાઇન ગાયબ હતી, જે મેં ઉમેરી દીધી છે!
        precision_score(y_test, knn_predict, zero_division=0),
        precision_score(y_test, nb_predict, zero_division=0)
    ],
    
    'Recall': [
        recall_score(y_test, lr_predict, zero_division=0),
        recall_score(y_test, dt_predict, zero_division=0),
        recall_score(y_test, rf_predict, zero_division=0),
        recall_score(y_test, svm_predict, zero_division=0),
        recall_score(y_test, knn_predict, zero_division=0),
        recall_score(y_test, nb_predict, zero_division=0)
    ],
    
    'F1-Score': [
        f1_score(y_test, lr_predict, zero_division=0),
        f1_score(y_test, dt_predict, zero_division=0),
        f1_score(y_test, rf_predict, zero_division=0),
        f1_score(y_test, svm_predict, zero_division=0),
        f1_score(y_test, knn_predict, zero_division=0),
        f1_score(y_test, nb_predict, zero_division=0)
    ]
}

results_df = pd.DataFrame(results)

# 4 decimal places
results_df = results_df.round(4)

print(results_df)
```
